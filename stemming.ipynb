Stemming is a text preprocessing technique in Natural Language Processing (NLP).

👉 It is the process of reducing a word to its root/base form by chopping off suffixes or prefixes. Root of words is known as lemma

The result may not always be a real dictionary word (sometimes it’s just a crude root).

🔹 Example:

Words:

"playing" → "play"

"played" → "play"

"plays" → "play"

"happiness" → "happi"

Here, "happi" is not a valid English word but is still considered a stem.

# Porter stemming


Uses a set of rules to iteratively chop suffixes like -ing, -ly, -ed, -ious, -ies, -ive, -es, -s, -ment.

Produces stems that may not always be dictionary words, but are consistent.

```python
from nltk.stem import PorterStemmer
```

```python
stemming=PorterStemmer()
```

```python
words=['eating' , 'eats' , 'eaten' , 'writing' , 'writes', 'programming', 'programs' , 'history', 'finally', 'finalize' ]
```

```python
for word in words:
    print(word+"--->"+stemming.stem(word))
```

```python
stemming.stem('congratulations')
```

```python
stemming.stem('reading')
```

# Regexpstemmer class
What is RegexpStemmer?

It is a rule-based stemmer that uses regular expressions to remove suffixes (or prefixes) from words.

Unlike Porter or Lancaster, which follow linguistic rules, RegexpStemmer simply applies a regex substitution.

This makes it customizable, but less intelligent.

```python
from nltk.stem import RegexpStemmer
```

```python
reg_stemmer = RegexpStemmer('ing|s$|e$|able$', min=4)
```

```python
reg_stemmer.stem('eating')
```

```python
reg_stemmer.stem('ingeating')
```

# Snowball stemmer

Also called the English Stemmer.

An improvement over Porter Stemmer (by the same author, Martin Porter).

More systematic and consistent, supports multiple languages (not just English).

Less aggressive than Lancaster, more refined than Porter.

```python
from nltk.stem import SnowballStemmer
```

```python
snowball= SnowballStemmer('english')
```

```python
for word in words:
    print(word+"-->"+snowball.stem(word))
```

```python
snowball.stem('congratulations')
```

```python
stemming.stem('fairly')
```

```python
snowball.stem('fairly') , snowball.stem('history'), snowball.stem('goes')
```

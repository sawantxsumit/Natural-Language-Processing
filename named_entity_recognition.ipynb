# Named entity recognition

```python
sentences='The eiffel tower was built from 1887 to 1889 as the entrance to the 1889 worlds fair in paris france it is named after the engineer gustave eiffel'
```

```python
import nltk
from nltk import word_tokenize, pos_tag , ne_chunk
```

```python
words=word_tokenize(sentences)
```

```python
tag_elemnts=pos_tag(words)
```

```python
nltk.download('maxent_ne_chunker_tab')
nltk.download('words')

```

```python
ne_chunk(tag_elemnts).draw()
```

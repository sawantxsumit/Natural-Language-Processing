```python
import pandas as pd
```

```python
df=pd.read_csv('data_processed.csv')
```

```python
df.head()
```

```python
df.drop(columns='Unnamed: 0', inplace=True)
```

```python
df.head()
```

```python
df['reviewText']=df['reviewText'].astype(str)
```

```python
df
```

```python
df['overall'].value_counts()
```

```python
df=df.head(10000)
```

```python
# Train test split
from sklearn.model_selection import train_test_split
```

```python
xtrain, xtest , ytrain , ytest = train_test_split(df['reviewText'], df['overall'] , test_size=0.25 , random_state=33)
```

```python
xtrain
```

```python
# Perform bag of words
from sklearn.feature_extraction.text import CountVectorizer
bow= CountVectorizer()
xtrain_bow= bow.fit_transform(xtrain).toarray()
xtest_bow= bow.transform(xtest).toarray()
```

```python
# Perform TFIDF
from sklearn.feature_extraction.text import TfidfVectorizer
tfidf= TfidfVectorizer()
xtrain_tfidf= tfidf.fit_transform(xtrain).toarray()
xtest_tfidf= tfidf.transform(xtest).toarray()
```

```python
xtrain_bow
```

```python
from sklearn.naive_bayes import GaussianNB
nb_model_bow=GaussianNB().fit(xtrain_bow, ytrain)
nb_model_tfidf=GaussianNB().fit(xtrain_tfidf, ytrain)
```

```python
from sklearn.metrics import confusion_matrix , accuracy_score, classification_report
```

```python
y_pred_bow= nb_model_bow.predict(xtest_bow)
```

```python
y_pred_tfidf= nb_model_tfidf.predict(xtest_tfidf)
```

```python
print('BOW accuracy :', accuracy_score(ytest , y_pred_bow))
```

```python
confusion_matrix(ytest , y_pred_bow)
```

```python
print('TF-IDF accuracy :', accuracy_score(ytest , y_pred_tfidf))
```

```python
confusion_matrix(ytest , y_pred_tfidf)
```

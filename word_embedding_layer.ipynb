```python
from tensorflow.keras.preprocessing.text import one_hot
```

```python
## sentences
sent= [
    'the glass of milk',
    'the glass of juice',
    'the cup of tea',
    'I am a good boy',
    'I am a good developer',
    'understand the meaning of words',
    'your videos are good'
]
```

```python
sent
```

```python
# vocab size
voc_size=10000
```

**One hot representation**

```python
onehot_repr= [one_hot(words, voc_size)for words in sent]
```

```python
print(onehot_repr)
```

**Word embedding representation**

```python
from tensorflow.keras.layers import Embedding , Input
from tensorflow.keras.preprocessing.sequence import pad_sequences
from tensorflow.keras.models import Sequential
```

```python
import numpy as np
```

This is used to set the length of all the sentences to the same length that is 8 in this scenario . This is called padding. It is achieved by adding zeroes in front or end
if zeroes are added in front it is called pre-padding and if the zeroes are added at the end it is called post-padding

It maked sure that the input size is fixed

```python
# Pre padding
sent_length=8
embedded_docs=pad_sequences(onehot_repr, padding='pre', maxlen=sent_length)
print(embedded_docs)
```

# Now we need to convert each and every value to vectors

```python
# Feature repr size (dimensions)
dim=10
```

```python
model= Sequential()
model.add(Input(shape=(sent_length,)))
model.add(Embedding(voc_size, 10))
model.compile('adam', 'mse')
```

```python
model.summary()
```

```python
print(model.predict(embedded_docs))
```

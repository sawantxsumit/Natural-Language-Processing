# Lemmatization
What is Lemmatization?

Lemmatization is the process of reducing a word to its base or dictionary form (called a lemma) using vocabulary + grammar rules.

Unlike stemming, which just chops off suffixes, lemmatization considers the meaning and context of the word.

The output is always a valid word in the language.

```python
from nltk.stem import WordNetLemmatizer
```

```python
lemmatizer= WordNetLemmatizer()
```

```python
import nltk
nltk.download('wordnet')
```

```python
lemmatizer.lemmatize('history')
```

```python
lemmatizer.lemmatize('goes')
```

```python
words=['eating' , 'eats' , 'eaten' , 'writing' , 'writes', 'programming', 'programs' , 'history', 'finally', 'finalize' ]
```

```python
for word in words:
    print(word+'-->'+lemmatizer.lemmatize(word, pos='v'))
```

```python
corpus= ''' The display property is an important CSS property for controlling layout. It specifies whether an HTML element is treated as a block or an inline element.

Every HTML element has a default display value, depending on what type of element it is. The default display value for most elements is block or inline.

The display property is used to change the default display behavior of HTML elements.
'''
```

```python
print(corpus)
```

```python
import nltk
nltk.download('punkt_tab')
```

```python
# Tokenization
from nltk.tokenize import sent_tokenize
```

```python
document=sent_tokenize(corpus)
```

```python
type(document)
```

```python
for sentence in document:
    print((sentence))
```

```python
# Tokenization
# para-> sentence
# sentence -> words

from nltk.tokenize import word_tokenize
```

```python
words=word_tokenize(corpus)
```

```python
for sentence in document:
    print(word_tokenize(sentence))
```

```python
from nltk.tokenize import wordpunct_tokenize
```

```python
wordpunct_tokenize(corpus)
# punctuations are treated as seperate words
```

```python
from nltk.tokenize import TreebankWordTokenizer
```

```python
tokenizer=TreebankWordTokenizer()
```

```python
tokenizer.tokenize(corpus)
#full stop is not treated as a separate word except the last full stop
```
